Lab Experiment 5- Outlier Detection and Treatment : Identifying outliers using IQR and Z-score methods and applying appropriate treatment

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 1:Load the Dataset

```python
df=pd.read_csv("/content/student_performance_dataset.csv")
print("First 5 rows of the dataset: ")
print(df.head())

print("\nStatistical Summary: ")
print(df.describe())
```

Step 2: Select numeric columns for outlier detection

```python
numeric_cols=["Study_Hours_per_Week","Attendance_Rate","Past_Exam_Scores","Final_Exam_Score"]
```

Step 2A: Artificially insert a few outliers

```python
df.loc[0,"Study_Hours_per_Week"]=120
df.loc[1,"Attendance_Rate"]=250
df.loc[2,"Past_Exam_Scores"]=-20
df.loc[3,"Final_Exam_Score"]=200
print("\nInserted a few artificial outlier values for demonstration.")
```

Step 3: Outlier detection using IQR Method

```python
print("\n-----IQR Method-----")
for col in numeric_cols:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[col]<lower) | (df[col]>upper)]
  print(f"{col} : {len(outliers)} outlier(s) found limits: {lower:.2f} to {upper:.2f}")
```

Step 4: Outlier Detection using Z-Score Method

```python
print("\n----------Z-Score Method----------")
for col in numeric_cols:
  z_scores=(df[col] - df[col].mean())/df[col].std()
  outliers=df[z_scores.abs()>3]
  print(f"{col}: {len(outliers)} outlier(s) found (|z|>3)")
```

Step 5: Treating Outliers

```python
df_treated=df.copy()
print("\n----------Treating outliers (Capping using IQR limits)----------")
for col in numeric_cols:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR

  df_treated[col]=df_treated[col].clip(lower,upper)
  print(f"{col}: value capped between {lower:.2f} and {upper:.2f}")

```

Step 6:Comparing Statistics before and after

```python
print("\nStatistics BEFORE Treatment:")
print(df[numeric_cols].describe())

print("\nStatistics AFTER Treatment:")
print(df_treated[numeric_cols].describe())
```

Step 7: Save the cleaned data

```python
df_treated.to_csv("student_performance_dataset_treated.csv",index=False)
print("\nTreated dataset saved as 'student_performance_dataset_treated.csv")
print("Program executed successfully.")
```
