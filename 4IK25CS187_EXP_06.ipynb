```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler,StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

Load Dataset and select the numeric features

```python
df=pd.read_csv("/content/Titanic-Dataset.csv")
df["Age"]=df['Age'].fillna(df['Age'].median())   #fill missing values
features=df[['Pclass','Age','SibSp','Parch','Fare']]
print('Original data\n',features.head())
```

Min-Max Normalization

```python
minmax=MinMaxScaler()
normalized=minmax.fit_transform(features)
print("\nNormalized data(first 5 rows):\n",normalized[:5].round(3))
print("Normalized range:",normalized.min(),"to",normalized.max())
```

Z score standarization

```python
scaler=StandardScaler()
standarized=scaler.fit_transform(features)
print("\nStandarized data (first 5 rows):\n",standarized[:5],round(3))
print("\nstandarized mean:",standarized.mean().round(4),"std:",standarized.std().round(4))
```

PCA for dimensionality reduction

```python
pca=PCA(n_components=2)
principal_components=pca.fit_transform(standarized)
print("\nShape beofore PCA:",standarized.shape,"| Shape after PCA:",principal_components.shape)
print("Explained variance ratio:",pca.explained_variance_ratio_.round(4))
print("Total variance captures by 2 componenets:",round(sum(pca.explained_variance_ratio_)*100,2),'%')
```

PCA Scatter plot

```python
colors=df["Survived"].map({1:"green",0:"red"})
plt.scatter(principal_components[:,0],principal_components[:,1],c=colors,alpha=0.6)
plt.scatter([],[],c="green",label="Survived")
plt.scatter([],[],c="red",label="Did Not survived")
plt.xlabel("PC1");plt.ylabel("PC2")
plt.title("PCA: 5 features reduced to 2")
plt.legend()
plt.show()
```
