Python for Data Analytics -Numpy , Pandas, Reading csv/json, indexing, filtering, grouping and aggregation

1.Basic Numpy Operations

```python
import numpy as np
import pandas as pd
marks=np.array([88,72,95,60,79])
print("Array: ",marks)
print("Mean: ",marks.mean())
print("Max: ",marks.max())
print("Min: ",marks.min())
```

2. Create a Sample CSV File

```python
data={
    "name":["Asha","Ravi","Meena","Kiran","Deepa"],
    "department":["CSE","ECE","CSE","ME","ECE"],
    "marks":[88,72,95,60,79]
}
df=pd.DataFrame(data)
df.to_csv("Students.csv",index=False)
df.to_json("Students.json",orient="records")
```

3. Reading CSV and JSON files

```python
print("\n----Reading CSV----")
df_csv=pd.read_csv("Students.csv")
print(df_csv)
```

```python
print("\n----Reading JSON----")
df_json=pd.read_json("Students.json")
print(df_json)
```

4. Indexing

```python
print("First marks: ",marks[0])
print("Second marks: ",marks[1])
print("Last marks: ",marks[-1])
```

```python
print("Marks column:")
print(df["marks"])
print("\nBranch and Marks column:")
print(df[["department","marks"]])
```

```python
print("Index 1 to 3: ",marks[1:4])
print("First three: ",marks[:3])
print("From index 2: ",marks[2:])
print("Every second element: ",marks[::2])
```

5. Filtering

```python
high_scorers = df[df["marks"]>90]
high_scorers
```

```python
print("Students from CSE:")
display(df[df["department"] == "CSE"])

print("\nStudents scoring 85 or above:")
display(df[df["marks"] >= 85])
```

6. Grouping And Aggregation

```python
print("Average marks per department:")
print(df.groupby("department")["marks"].mean())
```

```python
print("\nNumber of students per department:")
print(df.groupby("department")["name"].count())
```
